# Environment Check

This notebook verifies the environment setup for the AeroNetra research project.

```python
import importlib.metadata
import platform
import sys

print("Python Info:")
print(f"Version: {sys.version}")
print(f"Platform: {platform.platform()}")
print(f"Architecture: {platform.architecture()[0]}")
```

```python
print("\nPackage Versions:")
packages = [
    "numpy", "pandas", "opencv-python-headless", "matplotlib", 
    "Pillow", "scikit-learn", "ultralytics", "torch", "torchvision"
]
for pkg in packages:
    try:
        version = importlib.metadata.version(pkg)
        print(f"{pkg}: {version}")
    except importlib.metadata.PackageNotFoundError:
        print(f"{pkg}: NOT FOUND")
```

```python
import random

import numpy as np
import torch

print("\nDevice Availability:")
device = "cpu"
if torch.cuda.is_available():
    device = "cuda"
    print("CUDA is available.")
    print(f"Device: {torch.cuda.get_device_name(0)}")
elif torch.backends.mps.is_available():
    device = "mps"
    print("Apple MPS is available.")
else:
    print("Using CPU.")

print(f"\nSelected Device: {device}")
```

```python
def set_seed(seed=42):
    """Set random seeds for reproducibility."""
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
    print(f"Random seeds set to {seed}")

set_seed()
```

```python
import importlib.util

print("\nTesting Imports:")
if importlib.util.find_spec("cv2") is not None:
    print("OpenCV is available.")
else:
    print("OpenCV is not available.")

if importlib.util.find_spec("ultralytics") is not None:
    print("Ultralytics YOLO is available.")
    # Intentionally avoiding downloading weights here.
else:
    print("Ultralytics is not available.")
```
