# Model Evaluation (local)

Compare model performance on the shared VisDrone validation split.

Evaluation logic is not yet implemented in the library
(`src/aeronetra/evaluation/` is a stub). The reference GPU pipeline lives in
[`kaggle/03_model_evaluation.ipynb`](../kaggle/03_model_evaluation.ipynb).

Per [`docs/EXPERIMENT_GUIDE.md`](../docs/EXPERIMENT_GUIDE.md): always compare models on
the same validation split, record an `InferenceMetadata` for every run, and never
fabricate results.

```
import sys
from pathlib import Path

# Resolve the repository root regardless of where the notebook is launched,
# then expose the package source on the import path.
REPO_ROOT = next(
    (p for p in [Path.cwd(), *Path.cwd().parents] if (p / "pyproject.toml").exists()),
    Path.cwd(),
)
sys.path.insert(0, str(REPO_ROOT / "src"))
```

```
# Prerequisites for local evaluation.
weights_dir = REPO_ROOT / "outputs/models"
available = sorted(p.name for p in weights_dir.glob("*.pt")) if weights_dir.exists() else []

print(f"Repository root: {REPO_ROOT}")
print(f"Weights directory: {weights_dir}")
print(f"Trained weights found: {available or 'none'}")

if not available:
    print(
        "\nNo local weights yet. Train on Kaggle (kaggle/02_model_training.ipynb), "
        "download best.pt into outputs/models/, then evaluate on the validation split."
    )
```
